10. Read in the image ('Apollo-11-launch.jpg').

```python
img = cv2.imread("Apollo-11-launch.jpg")

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Apollo 11 Launch")
plt.axis("off")
plt.show()
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN

h, w = img.shape[:2]
text_size = cv2.getTextSize(text, font_face, 2, 2)[0]
x = (w - text_size[0]) // 2
y = h - 20

cv2.putText(img, text, (x, y), font_face, 2, (255, 255, 255), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

12. Draw a magenta rectangle that encompasses the launch tower and the rocket.

```python
# rect_color = magenta
rect_color = (255, 0, 255)

# Exact bounding box for both Launch Tower and Rocket
cv2.rectangle(img, (290, 200), (360, 630), rect_color, 3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

13. Display the final annotated image.

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Final Annotated Image")
plt.axis("off")
plt.show()
```

14. Read the image ('boy.jpg').

```python
img = cv2.imread("boy.jpg")

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Boy Image")
plt.axis("off")
plt.show()
```

15. Adjust the brightness of the image.

```python
matrix_ones = np.ones(img.shape, dtype="uint8")

bright_img = cv2.add(img, matrix_ones * 50)

plt.imshow(cv2.cvtColor(bright_img, cv2.COLOR_BGR2RGB))
plt.title("Brightened Image")
plt.axis("off")
plt.show()
```

16. Create brighter and darker images.

```python
#img_brighter = cv2.add(img, matrix)
#img_darker = cv2.subtract(img, matrix)
# YOUR CODE HERE
matrix = np.ones(img.shape, dtype="uint8") * 50

img_brighter = cv2.add(img, matrix)
img_darker = cv2.subtract(img, matrix)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB))
plt.title("Brighter Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB))
plt.title("Darker Image")
plt.axis("off")

plt.show()
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB))
plt.title("Darker Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB))
plt.title("Brighter Image")
plt.axis("off")

plt.show()
```

18. Modify the image contrast.

```python
matrix1 = np.ones(img.shape, dtype="uint8")
matrix2 = np.ones(img.shape, dtype="uint8")

img_higher1 = cv2.multiply(img, matrix1, scale=1.1)
img_higher2 = cv2.multiply(img, matrix2, scale=1.2)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img_higher1, cv2.COLOR_BGR2RGB))
plt.title("Contrast 1.1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_higher2, cv2.COLOR_BGR2RGB))
plt.title("Contrast 1.2")
plt.axis("off")

plt.show()
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
matrix = np.ones(img.shape, dtype="uint8") * 50
img_lower = cv2.subtract(img, matrix)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img_lower, cv2.COLOR_BGR2RGB))
plt.title("Lower Contrast")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(img_higher1, cv2.COLOR_BGR2RGB))
plt.title("Higher Contrast")
plt.axis("off")

plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
img = cv2.imread("boy.jpg")

b, g, r = cv2.split(img)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(b, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(g, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(r, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
merged_img = cv2.merge([b, g, r])

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(merged_img, cv2.COLOR_BGR2RGB))
plt.title("Merged Image")
plt.axis("off")

plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(hsv_img)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(h, cmap="gray")
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(s, cmap="gray")
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(v, cmap="gray")
plt.title("Value Channel")
plt.axis("off")

plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged_hsv = cv2.merge([h, s, v])
merged_img = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(merged_img, cv2.COLOR_BGR2RGB))
plt.title("Merged HSV Image")
plt.axis("off")

plt.show()
```
